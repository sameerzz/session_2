```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage

load_dotenv()

llm = ChatOpenAI(
    model="google/gemini-2.5-flash",
    api_key=os.environ["OPEN_ROUTER_API_KEY"],
    base_url="https://openrouter.ai/api/v1",
    temperature=0,
)
```

```python
from tools import sign_in, tools

email = input("Enter your shopping email: ").strip()
sign_in(email)

# Map a tool name to the function we can execute.
tool_repo = {tool.name: tool for tool in tools}

# Give the model the tools' descriptions and argument schemas.
llm_with_tools = llm.bind_tools(tools)
```

```python
import tools as shop_tools

# Read the header saved by sign_in(). Use the module's current value.
authorization = shop_tools.headers["Authorization"]
login_token = authorization.removeprefix("Bearer ")

# Show the shape, keeping most of the real token hidden.
print("Login token:", login_token[:6] + "..." + login_token[-4:])
```

```python
import httpx

response = httpx.get(
    f"{shop_tools.API_URL}/cart",
    headers={"Authorization": f"Bearer {login_token}"},
    timeout=30,
)
response.raise_for_status()
response.json()
```

```python
# Keep shared behaviour here; tool-specific details live in docstrings.
sysmsg = """
You are a helpful Hopscotch shopping assistant.
Use the available tools when needed and follow their descriptions.
Complete prerequisite calls before requesting dependent actions.
Only take actions the user explicitly requests.
Never invent customer data or claim success without a successful tool result.
If a request is ambiguous, explain what information is missing.
Avoid unnecessary tool calls and give one clear final answer.
"""
```

```python
system_message = SystemMessage(content=sysmsg)
user_query = input("Enter your message?")
user_message = HumanMessage(content=user_query)
messages_to_send = [system_message, user_message]

response = llm_with_tools.invoke(messages_to_send)
while(response.tool_calls):
  # Keep this AI request before adding its tool results.
  messages_to_send.append(response)
  for tool_call in response.tool_calls:
    tool_name = tool_call['name']
    tool_output = tool_repo[tool_name].invoke(tool_call)
    messages_to_send.append(tool_output)
  response = llm_with_tools.invoke(messages_to_send)
messages_to_send.append(response)
print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

# Run this cell once to start a conversation; rerun it to clear history.
history = []

prompt = ChatPromptTemplate.from_messages([
    ("system", sysmsg),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{query}"),
])
```

```python
user_query = input("Enter your message?")
prompt_value = prompt.invoke({"history": history, "query": user_query})
prompt_value.to_messages()
```

```python
chain = prompt | llm_with_tools

response = chain.invoke({"history": history, "query": user_query})
response
```
